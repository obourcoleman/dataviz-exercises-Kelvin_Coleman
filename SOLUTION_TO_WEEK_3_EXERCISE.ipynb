```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

```python
print(df.columns.tolist())
```

```python
# Choose the country to highlight
highlighted_country = "Germany"

# Create an empty figure
fig = go.Figure()

# Plot all countries in grey
for country in df["Country"].unique():
    if country != highlighted_country:
        country_data = df[df["Country"] == country].sort_values("Year")

        fig.add_trace(go.Scatter(
            x=country_data["Year"],
            y=country_data["CO2_per_capita"],
            mode="lines",
            line=dict(color="#EEEEEE", width=1),
            showlegend=False,
            hoverinfo="skip"
        ))

# Add the highlighted country
country_data = df[df["Country"] == highlighted_country].sort_values("Year")

fig.add_trace(go.Scatter(
    x=country_data["Year"],
    y=country_data["CO2_per_capita"],
    mode="lines+text",
    line=dict(color="red", width=3),
    text=[None] * (len(country_data) - 1) + [highlighted_country],
    textposition="middle right",
    showlegend=False
))

# Add chart title and labels
fig.update_layout(
    title=f"CO2 per capita: {highlighted_country} against all other countries",
    xaxis_title="Year",
    yaxis_title="CO2 per capita",
    template="plotly_white"
)

fig.show()
```

```python
# Select Asian countries
asia_df = df[df["Region"] == "Asia"]

# Choose the country to highlight
highlighted_country = "China"

# Create an empty figure
fig = go.Figure()

# Plot all other Asian countries in grey
for country in asia_df["Country"].unique():
    if country != highlighted_country:
        country_data = asia_df[asia_df["Country"] == country].sort_values("Year")

        fig.add_trace(go.Scatter(
            x=country_data["Year"],
            y=country_data["CO2_Mt"],
            mode="lines",
            line=dict(color="#EEEEEE", width=1),
            showlegend=False,
            hoverinfo="skip"
        ))

# Plot the highlighted country
country_data = asia_df[asia_df["Country"] == highlighted_country].sort_values("Year")

fig.add_trace(go.Scatter(
    x=country_data["Year"],
    y=country_data["CO2_Mt"],
    mode="lines+text",
    line=dict(color="red", width=3),
    text=[None] * (len(country_data) - 1) + [highlighted_country],
    textposition="middle right",
    showlegend=False
))

# Format the chart
fig.update_layout(
    title="China dominates CO2 emissions among Asian countries",
    xaxis_title="Year",
    yaxis_title="CO2 emissions (Mt)",
    template="plotly_white"
)

fig.show()
```
