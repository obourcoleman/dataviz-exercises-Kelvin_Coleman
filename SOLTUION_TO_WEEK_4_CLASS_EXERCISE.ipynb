```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

```python

# Task 1 — Scatter: GDP per capita vs Life Expectancy

# Filter Europe for 2002 and 2007

import plotly.express as px
import plotly.graph_objects as go

# Filter Europe for 2002 and 2007
df_task1 = df[
    (df['continent'] == 'Europe') &
    (df['year'].isin([2002, 2007]))
].copy()

# Create scatter plot
fig = px.scatter(
    df_task1,
    x='gdpPercap',
    y='lifeExp',
    color='year',
    hover_name='country',
    log_x=True,
    title='Europe: GDP and Life Expectancy (2002–2007)',
    labels={
        'gdpPercap': 'GDP per Capita',
        'lifeExp': 'Life Expectancy',
        'year': 'Year'
    },
    color_discrete_map={
        2002: 'blue',
        2007: 'orange'
    }
)

# Draw lines connecting 2002 and 2007
for country in df_task1['country'].unique():

    data = df_task1[
        df_task1['country'] == country
    ].sort_values('year')

    if len(data) == 2:
        fig.add_trace(
            go.Scatter(
                x=data['gdpPercap'],
                y=data['lifeExp'],
                mode='lines',
                line=dict(color='gray', width=1),
                showlegend=False,
                hoverinfo='skip'
            )
        )

# Update marker size
fig.update_traces(
    marker=dict(size=9),
    selector=dict(mode='markers')
)

fig.show()

fig.show()
```

```python

import plotly.express as px

# Filter data for 2007
df_2007 = df[df['year'] == 2007].copy()

# Highlight Europe and grey out other continents
df_2007['highlight'] = df_2007['continent'].apply(
    lambda x: 'Europe' if x == 'Europe' else 'Other'
)

# Create bubble chart
fig = px.scatter(
    df_2007,
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='highlight',
    hover_name='country',
    log_x=True,
    size_max=60,
    color_discrete_map={
        'Europe': 'blue',
        'Other': 'lightgrey'
    },
    title='European Countries Have Higher GDP and Life Expectancy',
    labels={
        'gdpPercap': 'GDP per Capita',
        'lifeExp': 'Life Expectancy',
        'pop': 'Population',
        'highlight': 'Continent'
    }
)

# Add annotation highlighting Europe
fig.add_annotation(
    x=30000,
    y=80,
    text='Europe: Higher GDP and Life Expectancy',
    showarrow=True,
    arrowhead=2,
    arrowcolor='blue',
    ax=-120,
    ay=-60,
    bgcolor='white',
    bordercolor='blue',
    borderwidth=1,
    font=dict(size=12, color='blue')
)

# Improve layout
fig.update_layout(
    width=1100,
    height=650,
    legend_title_text='Continent'
)

# Display graph
fig.show()

```
