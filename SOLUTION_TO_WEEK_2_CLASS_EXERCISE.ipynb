```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# World Happiness Report 2023 — representative data
# Source: https://www.kaggle.com/datasets/ajaypalsinghlo/world-happiness-report-2023

df = pd.read_csv('world_happiness_2023.csv')
df.columns = ['Country','Region','Happiness_Score','GDP','Social_Support',
              'Life_Expectancy','Freedom','Generosity','Corruption']


print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
print(df.head())
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Explore the dataset before you start
print("Regions in dataset:")
print(df['Region'].value_counts())
print("\nScore range:", df['Happiness_Score'].min(), "–", df['Happiness_Score'].max())
print("\nBottom 10 countries:")
print(df.nsmallest(10, 'Happiness_Score')[['Country','Region','Happiness_Score']])
```

```python
# Task 1: Regional comparison bar chart
# -------------------------------------

# Step 1: Compute average happiness score by region
region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))  # sort for horizontal bar

print(region_avg)

# Step 2: Build your chart
# YOUR CODE HERE
fig = px.bar(
    region_avg,
    x='Happiness_Score',
    y='Region',
    orientation='h',
    title='Western Europe Leads the World in Happiness',
    labels={
        'Happiness_Score': 'Average Happiness Score',
        'Region': ''
    },
    text='Happiness_Score'
)

fig.update_traces(
    texttemplate='%{text:.2f}',
    textposition='outside'
)

fig.update_layout(
    xaxis=dict(range=[0, None]),
    showlegend=False
)

fig.show()
```

```python
# Task 2: Top 8 vs. Bottom 8 contrast
# ------------------------------------

# Step 1: Get top and bottom countries
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
print(f"Global average: {global_avg:.2f}")

# Step 2: Build your chart
# YOUR CODE HERE
fig = px.bar(
    combined,
    x='Happiness_Score',
    y='Country',
    color='Group',
    orientation='h',
    title='The Happiness Gap Between the Happiest and Least Happy Countries',
    color_discrete_map={
        'Top 8': 'green',
        'Bottom 8': 'red'
    },
    labels={
        'Happiness_Score': 'Happiness Score',
        'Country': ''
    }
)

# Global average reference line
fig = px.bar(
    combined,
    x='Happiness_Score',
    y='Country',
    color='Group',
    orientation='h',
    title='The Happiness Gap Between the Happiest and Least Happy Countries',
    color_discrete_map={
        'Top 8': 'green',
        'Bottom 8': 'red'
    },
    labels={
        'Happiness_Score': 'Happiness Score',
        'Country': ''
    }
)

# Global average reference line
fig.add_vline(
    x=global_avg,
    line_dash='dash',
    annotation_text=f'Global Average: {global_avg:.2f}',
    annotation_position='top'
)

# Clear separator between the two groups
fig.add_hline(
    y=7.5,
    line_width=3
)

fig.update_layout(
    xaxis=dict(range=[0, None]),
    legend_title_text='Group'
)

fig.show()
```

```python
#TASK 3
regions = [
    'Western Europe',
    'Latin America',
    'East Asia',
    'Sub-Saharan Africa',
    'South Asia'
]

region_factors = (
    df[df['Region'].isin(regions)]
    .groupby('Region')[['GDP', 'Freedom']]
    .mean()
    .reset_index()
)

print(region_factors)

fig = px.bar(
    region_factors,
    x='Region',
    y=['GDP', 'Freedom'],
    barmode='group',
    title='Western Europe Leads in Both Economic Strength and Freedom',
    labels={
        'value': 'Average Score',
        'variable': 'Factor',
        'Region': ''
    },
    color_discrete_map={
        'GDP': 'steelblue',
        'Freedom': 'orange'
    }
)

fig.update_layout(
    yaxis=dict(range=[0, None]),
    legend_title_text='Factor'
)

fig.show()
```
